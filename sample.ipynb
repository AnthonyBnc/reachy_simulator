```python
import gymnasium as gym, io
from PIL import Image
import ipywidgets as widgets
from IPython.display import display

env = gym.make("reachy_sim:ReachyReach-v0", render_mode="rgb_array")
obs, info = env.reset()
image_widget = widgets.Image(format="png")
display(image_widget)

for i in range(200):
    buf = io.BytesIO()
    Image.fromarray(env.render()).save(buf, format="png")
    image_widget.value = buf.getvalue()

    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    if info["is_success"]:
        print("Yay!")
        break
    if terminated or truncated:
        obs, info = env.reset()
env.close()
```
